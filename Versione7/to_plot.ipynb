## Gaussian DATA

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# === 1. Carica il file CSV ===
path = "/Users/alessandromazzeo/Desktop/UCL/CODE/Versione7/"
df = pd.read_csv(path+"frobenius_results_gaussian.csv")

# === 2. Elenco dei metodi ===
methods = ['Sample', 'RIE', 'RIE_IW', 'Clipped', 'Kendall', 'TMFG']
selected_methods = ["TMFG", "Kendall", "Clipped"]

# === 3. Palette coerente ===
palette = sns.color_palette("tab10", n_colors=len(methods))
color_map = {method: palette[i] for i, method in enumerate(methods)}

# === 4. Crea i dizionari con array numpy ===
mean_dict = {}
std_dict = {}
q_dict = {}

for method in methods:
    df_m = df[df["method"] == method].sort_values("q")
    q_dict[method]    = df_m["q"].to_numpy()
    mean_dict[method] = df_m["mean"].to_numpy()
    std_dict[method]  = df_m["std"].to_numpy()

# === 5. Canvas con due subplot NON condividendo y ===
sns.set(style="whitegrid", font_scale=1.2)
fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=False)

# --- Plot a sinistra: tutti i metodi, zoom su x ---
ax_left = axes[0]
for method in methods:
    q = q_dict[method]
    mean = mean_dict[method]
    std = std_dict[method]
    ax_left.plot(q, mean, label=method, color=color_map[method], marker='o')
    ax_left.fill_between(q, mean - 1*std, mean + 1*std, alpha=0.2, color=color_map[method])

ax_left.set_xlabel(r"$q = N / T$", fontsize=13)
ax_left.set_ylabel("Relative Frobenius Distance", fontsize=13)
ax_left.set_xlim(0.5, 1.3)  # Zoom su q
ax_left.set_yscale("log")   # Scala log
ax_left.set_title("All Methods (Zoomed in $q$)", fontsize=14)
ax_left.legend(loc="upper left", fontsize=9)

# --- Plot a destra: solo TMFG, Clipped, Kendall su x completo, y indipendente ---
ax_right = axes[1]
for method in selected_methods:
    q = q_dict[method]
    mean = mean_dict[method]
    std = std_dict[method]
    ax_right.plot(q, mean, label=method, color=color_map[method], marker='o')
    ax_right.fill_between(q, mean - 1*std, mean + 1*std, alpha=0.2, color=color_map[method])

ax_right.set_xlabel(r"$q = N / T$", fontsize=13)
ax_right.set_title("Selected Methods (Full Range)", fontsize=14)
ax_right.legend(loc="upper left", fontsize=9)

# Y non logaritmico e con scala propria (opzionale, o anche log se vuoi)
# ax_right.set_yscale("log")  # solo se vuoi anche a destra log
# ax_right.set_ylim(...)      # imposta se vuoi forzare limiti visivi

plt.tight_layout()
#plt.savefig("Frobenius_Comparison_SideBySide_Gaussian.pdf")
plt.show()
```
